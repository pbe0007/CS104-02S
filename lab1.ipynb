```python
#get a number from the user
num0 = int(input("Give a number greater than 20: "))

#turn the input into a binary and hex representation, this is done here for easier manipulation in the output
test1 = str(bin(num0))
test2 = str(hex(num0))

#output both conversions in a nice, easy-to-read format (removing prefix)
print(str(num0) + " in binary is " + test1[2:])
print(", in hex it is " + test2[2:])

#convert both back into base-10
print("just to check...")

print(int(test1,2))
print(int(test2,16))
```
